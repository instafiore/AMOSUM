# Load

`results_tsv/*.tsv` -> `results_csv/results.csv`: the runs, cleaned, with the metrics (`optimum`, `win`, `score`).
The metrics compare the solvers kept here, so filter **before** `add_metrics`.

```python
%load_ext autoreload
%autoreload 2
import pandas as pd
import lib
pd.set_option("display.max_columns", 50)
```

## Read the TSV files

```python
files = None                      # None: every results_tsv/*.tsv; or a list, e.g. [lib.TSV_DIR / "....tsv"]
raw = lib.load_tsv(files)
print(len(raw), "rows from", raw["tsv"].nunique(), "file(s)")
raw.head()
```

## Clean (types, names, duplicated runs)

```python
df = lib.clean(raw)
df.groupby(["bench", "solver"]).size().unstack(fill_value=0)
```

## Filter

Lists of `{column: regex}`; e.g. `drop = [{"bench": "^TSP$"}]`.

```python
keep = []
drop = []
df = lib.filter_df(df, keep, drop)
missing = lib.check_complete(df)
print(len(missing), "instances without a run of every solver")
missing.head()
```

## Metrics

```python
df = lib.add_metrics(df)
df[["bench", "instance", "solver", "status", "cost", "best", "optimum", "win", "score"]].head(10)
```

## Save

```python
lib.CSV_DIR.mkdir(exist_ok=True)
csv_path = lib.CSV_DIR / "results.csv"
df.to_csv(csv_path, index=False)
csv_path
```
