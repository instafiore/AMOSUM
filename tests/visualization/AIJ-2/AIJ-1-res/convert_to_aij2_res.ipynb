# Extract

```python
import os
import pandas as pd

df = pd.read_csv("amosum_experiments_clown-version.csv", sep=",", header="infer")
df.drop(columns=["solver","type","version","lazy","lang","minimality","solved","par1","par10","status_1","status_2"], inplace=True)
df
```

```python
df["problem"].unique()
```

# Remove WASP and Clingo version

```python
solver_mask_wasp_eo = df["executable"].str.contains("wasp-plain-eo")
solver_mask_clingo_eo = df["executable"].str.contains("clingo-plain-eo")

solver_mask_wasp_amo = df["executable"].str.contains("wasp-plain-amo")
solver_mask_clingo_amo = df["executable"].str.contains("clingo-plain-amo")

problem_mask_eo =  df["problem"].str.contains(r"(Knapsack|GroupAssignment)")
problem_mask_amo =  df["problem"].str.contains(r"Nurse")

df_cleaned = df.copy()
for solver_eo in [solver_mask_wasp_eo, solver_mask_clingo_eo]:
    df_cleaned.drop(index=df[solver_eo & problem_mask_eo].index, inplace=True)

for solver_amo in [solver_mask_wasp_amo, solver_mask_clingo_amo]:
    df_cleaned.drop(index=df[solver_amo & problem_mask_amo].index, inplace=True)
```

# Remove AMOWASP and AMOCLINGO version

```python
solver_mask_amowasp_eo = df["executable"].str.contains("wasp-amosum-eo")
solver_mask_amoclingo_eo = df["executable"].str.contains("clingo-amosum-eo")

solver_mask_amowasp_amo = df["executable"].str.contains("wasp-amosum-amo")
solver_mask_amoclingo_amo = df["executable"].str.contains("clingo-amosum-amo")

for solver_eo in [solver_mask_amowasp_eo, solver_mask_amoclingo_eo]:
    df_cleaned.drop(index=df[solver_eo & problem_mask_eo].index, inplace=True)

for solver_amo in [solver_mask_amowasp_amo, solver_mask_amoclingo_amo]:
    df_cleaned.drop(index=df[solver_amo & problem_mask_amo].index, inplace=True)
```

```python
df_cleaned[df["executable"].str.contains("B-")]["executable"].unique()
```

```python
df_cleaned.drop(index=df_cleaned[df_cleaned["executable"].str.contains(r"^C-")].index, inplace=True)
df_cleaned.drop(index=df_cleaned[df_cleaned["executable"].str.contains(r"^D-")].index, inplace=True)
df_cleaned.loc[:, "executable"] = df_cleaned.loc[:, "executable"].str.replace(r"wasp-plain","wasp")
df_cleaned.loc[:, "executable"] = df_cleaned.loc[:, "executable"].str.replace(r"clingo-plain","clingo")
df_cleaned.loc[:, "executable"] = df_cleaned.loc[:, "executable"].str.replace(r"wasp-amosum","amowasp")
df_cleaned.loc[:, "executable"] = df_cleaned.loc[:, "executable"].str.replace(r"clingo-amosum","amoclingo")

# IJCAI
df_cleaned.loc[:, "executable"] = df_cleaned.loc[:, "executable"].str.replace(r"ijcai-(amoclingo|amowasp)-(amo|eo)-(py|c)",r"\1-\2-base-lg=\3", regex=True)

# Improved Inference and Activation progation
df_cleaned.loc[:, "executable"] = df_cleaned.loc[:, "executable"].str.replace(r"A-(amoclingo|amowasp)-(amo|eo)-(py|c)",r"\1-\2-lg=\3-r=ijcai-l=f", regex=True)

# Improved reason and lazy
df_cleaned.loc[:, "executable"] = df_cleaned.loc[:, "executable"].str.replace(r"B-(amoclingo|amowasp)-(amo|eo)-(py|c)$",r"\1-\2-lg=\3-r=nomin-l=f", regex=True)
df_cleaned.loc[:, "executable"] = df_cleaned.loc[:, "executable"].str.replace(r"B-(amoclingo|amowasp)-(amo|eo)-(py|c)-hybrid$",r"\1-\2-lg=\3-r=nomin-l=h", regex=True)
df_cleaned.loc[:, "executable"] = df_cleaned.loc[:, "executable"].str.replace(r"B-(amoclingo|amowasp)-(amo|eo)-(py|c)-full$",r"\1-\2-lg=\3-r=nomin-l=t", regex=True)

# Minimal reason on the fly and lazy
df_cleaned.loc[:, "executable"] = df_cleaned.loc[:, "executable"].str.replace(r"fixedD-(amoclingo|amowasp)-(amo|eo)-(py|c)$",r"\1-\2-lg=\3-r=minfly-l=f", regex=True)
df_cleaned.loc[:, "executable"] = df_cleaned.loc[:, "executable"].str.replace(r"fixedD-(amoclingo|amowasp)-(amo|eo)-(py|c)-hybrid$",r"\1-\2-lg=\3-r=minfly-l=h", regex=True)
df_cleaned.loc[:, "executable"] = df_cleaned.loc[:, "executable"].str.replace(r"fixedD-(amoclingo|amowasp)-(amo|eo)-(py|c)-full$",r"\1-\2-lg=\3-r=minfly-l=t", regex=True)

df_cleaned["executable"].unique()
```

# Computing Metrics

```python
# solved,par1,par10
import numpy as np


df_cleaned.loc[:, "solved"] = df_cleaned["status"] == "complete"
df_cleaned.loc[:, "par1"] = np.where(
    df_cleaned["status"] == "complete",
    df_cleaned["real"],
    1200
)
df_cleaned.loc[:, "par10"] = np.where(
    df_cleaned["status"] == "complete",
    df_cleaned["real"],
    1200*10
)
```

```python
df_cleaned["executable"].unique()
```

# Df AMO

```python
df_amo = df_cleaned.copy()
df_amo.drop(index=df_cleaned[df_cleaned["executable"].str.contains(r"(-eo-|-eo$)")].index, inplace=True)
df_amo.loc[:, "executable"] = df_amo["executable"].str.replace("-amo-","-")
df_amo.loc[:, "executable"] = df_amo["executable"].str.replace("-amo$","", regex=True)
df_amo["executable"].unique()
```

## Pivot AMO

```python
pivot_amo = df_amo.pivot_table(index=["executable"],columns=["problem"], values=["solved","par1", "par10"], aggfunc={"solved":"sum", "par1": "mean", "par10": "mean"}, margins=True, margins_name='Total', fill_value=0)
pivot_amo = pivot_amo.reorder_levels([1,0], axis=1).sort_index(axis=1)
pivot_amo = pivot_amo.drop(index="Total")
# pivot_amo.loc[:, "solved"].astype(int)
for col in pivot_amo.columns[(pivot_amo.columns.get_level_values(1) == "solved")]:
    pivot_amo[col] = pivot_amo.loc[:, col].astype(int)
pivot_amo
```

# Df EO

```python
df_eo = df_cleaned.copy()
df_eo.drop(index=df_cleaned[df_cleaned["executable"].str.contains("(-amo-|-amo$)")].index, inplace=True)
df_eo.loc[:, "executable"] = df_eo["executable"].str.replace("-eo-","-")
df_eo.loc[:, "executable"] = df_eo["executable"].str.replace(r"-eo$","")
df_eo
```

## Pivot EO

```python
pivot_eo = df_eo.pivot_table(index=["executable"],columns=["problem"], values=["solved","par1", "par10"], aggfunc={"solved":"sum", "par1": "mean", "par10": "mean"}, margins=True, margins_name='Total', fill_value=0)
pivot_eo = pivot_eo.reorder_levels([1,0], axis=1).sort_index(axis=1)
pivot_eo = pivot_eo.drop(index="Total")

for col in pivot_eo.columns[(pivot_eo.columns.get_level_values(1) == "solved")]:
    pivot_eo[col] = pivot_eo.loc[:, col].astype(int)
pivot_eo
```
